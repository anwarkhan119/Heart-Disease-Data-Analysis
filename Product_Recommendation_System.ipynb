```python
from datasets import load_dataset

url = "https://huggingface.co/datasets/McAuley-Lab/Amazon-Reviews-2023/resolve/main/raw_meta_All_Beauty/full-00000-of-00001.parquet"

dataset = load_dataset("parquet", data_files=url, split="train")

print(dataset)
```

```python
products = df[
    [
        "title",
        "average_rating",
        "rating_number",
        "features",
        "description",
        "price",
        "store",
        "categories"
    ]
].copy()

products.head()
```

```python
products.shape
```

```python
products["text"] = (
    products["title"].fillna("") + " " +
    products["features"].fillna("").astype(str) + " " +
    products["description"].fillna("").astype(str) + " " +
    products["categories"].fillna("").astype(str)
)

products[["title", "text"]].head()
```

```python
products["text"] = products["text"].str.lower()
products["text"] = products["text"].str.replace(r"\s+", " ", regex=True)

products[["title", "text"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=20000
)

tfidf_matrix = tfidf.fit_transform(products["text"])

print(tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_products(product_name, n=5):

    # Find products matching the search term
    matches = products[
        products["title"].str.contains(
            product_name,
            case=False,
            na=False
        )
    ]

    # Check if the product exists
    if matches.empty:
        print("Product not found.")
        return

    # Get the index of the first matching product
    idx = matches.index[0]

    # Calculate similarity with all products
    scores = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    # Sort products by similarity score
    similar_indices = scores.argsort()[::-1]

    # Remove the selected product itself
    similar_indices = [
        i for i in similar_indices
        if i != idx
    ][:n]

    # Create recommendation result
    result = products.iloc[similar_indices][
        [
            "title",
            "average_rating",
            "price",
            "store"
        ]
    ].copy()

    # Add similarity score
    result["similarity_score"] = scores[similar_indices]

    return result
```

```python
def recommend_products(product_name, n=5):

    # Find products matching the search term
    matches = products[
        products["title"].str.contains(
            product_name,
            case=False,
            na=False
        )
    ]

    if matches.empty:
        print("Product not found.")
        return

    # Get the first matching product
    idx = matches.index[0]

    # Calculate similarity with all products
    similarity_scores = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    # Create a copy of the product data
    result = products.copy()

    # Add similarity score
    result["similarity_score"] = similarity_scores

    # Normalize ratings between 0 and 1
    result["rating_score"] = result["average_rating"] / 5

    # Create a combined recommendation score
    result["recommendation_score"] = (
        0.7 * result["similarity_score"] +
        0.3 * result["rating_score"]
    )

    # Remove the selected product
    result = result.drop(index=idx)

    # Sort by recommendation score
    result = result.sort_values(
        by="recommendation_score",
        ascending=False
    )

    # Return top recommendations
    return result.head(n)[
        [
            "title",
            "price",
            "average_rating",
            "similarity_score",
            "recommendation_score"
        ]
    ]
```
